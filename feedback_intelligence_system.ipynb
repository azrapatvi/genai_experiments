#### 2.Define tempalte :
the context builder this will build the context and context contains human prompt +system prompt this context then provided to the llm to get the response.

- langchain_core: contains all the primitives like template (context builder), output parser

- langchain: contains the agentic harness

- langgraph: used to build complex custom ai agents and workflow

- langsmith: provides observablity into ai applicatiosn you are building

````python
from langchain_core.prompts import ChatPromptTemplate

template=ChatPromptTemplate(
    messages=[
        ("system","""
          you are a strict json generator.
          analyze the following customer feedback.

          return STRICT JSON with:
          -sentiment (positive,neutral,negative)
          -category (product,delivery,support,pricing,other)
          -issue (short summary)

          only return JSON.No explaination.
        """), # system prompt
        ("human","""Feedback: ```{feedback}```""") #human prompt
    ]
)
````

````python
template.invoke({"feedback":"i love this product"})
````

````python
# pydantic is used to check and validate the data
# means its like hey my data has some rules that you need to follow in order to run this

# means this allows you to let the user enter the correct info
# ex: lets say there is a form that accepts name age adn email
# so
# name=string
# age=int
# email=string

# means
# name="azra"
# age=20
# email="azra@gmail.com"

# you cant enter like this
# name="azra"
# age="twenty"
# email="azra@gmail.com"

# here age is not correct so pydantic willl through an error(age must be int or soemhting) if you define it using basemodel


from pydantic import BaseModel

class Student(BaseModel):
  name:str
  age:int
  email:str

# means we are saying
# Student
# │
# ├── name  → must be a string
# ├── age   → must be an integer
# └── email → must be a string

student=Student(name="azra",age=20,email='azra@gmail.com')

print(student)
print(type(student))
````

````python
student = Student(
    name="Azra",
    age="hello",
    email="azra@gmail.com"
)
print(student)
print(type(student))

# so here will gett validation error because of pydantic...
# pydantic defines the rules or used for vlaidation
````

````python
from pydantic import BaseModel
from typing import Literal

class feedbackoutput(BaseModel):
  sentiment:Literal["positive","negative","neutral"]
  category:Literal["product","delivery","support","pricing","other"]
  issue:str

# so the strucutre of the output i wantt from the parser is ready using pydantic
````

````python
from langchain_core import output_parsers


# so we need to provide that info to ouput parse
from langchain_core.output_parsers import PydanticOutputParser

json_parser=PydanticOutputParser(pydantic_object=feedbackoutput)
````

````python
# creating a chain or connection betweeen this template and chat model

chain= template | gpt_llm | json_parser
````
